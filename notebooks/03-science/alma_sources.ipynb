### Reprojection

* According to https://reproject.readthedocs.io/en/stable/celestial.html, resampling assumes that the image is in surface brigtness units.
* SFR image is in sfr/pixel, and want sfr/sr, so we divide by sr_per_pixel
* then reproject
* testing a few algorithms - reproject_interp and reproject_adaptive with conserve_flux = True
* output of reproject should be in SB
* to convert output image to flux, then we need to multiply by the sr/pix

````python
def reproject_mask_to_reference(infile, reffile, outfile, overwrite=False):
    from pathlib import Path
    from astropy.io import fits
    from reproject import reproject_interp
    import numpy as np

    infile = Path(infile)
    reffile = Path(reffile)
    outfile = Path(outfile)

    if outfile.exists() and not overwrite:
        print(f"Using existing reprojected mask: {outfile}")
        return outfile

    data, hdr = fits.getdata(infile, header=True)
    ref_data, ref_hdr = fits.getdata(reffile, header=True)

    new_data, footprint = reproject_interp(
        (data.astype(float), hdr),
        ref_hdr,
        shape_out=ref_data.shape,
        order="nearest-neighbor",
    )

    mask_bool = (new_data > 0.5) | ~np.isfinite(new_data) | (footprint == 0)

    out_hdr = ref_hdr.copy()
    out_hdr["REPROJ"] = (True, "Mask reprojected to reference grid")
    out_hdr["REPSRC"] = (infile.name, "Source mask")
    out_hdr["REPREF"] = (reffile.name, "Reference image")
    out_hdr["BUNIT"] = ("mask", "1=masked/bad, 0=good")

    fits.PrimaryHDU(data=mask_bool.astype(np.uint8), header=out_hdr).writeto(
        outfile,
        overwrite=True,
    )

    print(f"Wrote {outfile}")

    return outfile


def _get_first_existing_attr(obj, names):
    for name in names:
        if hasattr(obj, name):
            return getattr(obj, name)

    available = sorted(obj.__dict__.keys())
    raise AttributeError(
        "Could not find any of these attributes:\n"
        f"{names}\n\n"
        "Available attributes include:\n"
        f"{available}"
    )

def get_hapy_ellipse_geometry(e, x0=None, y0=None, ba=None, pa_deg=None):
    """
    Extract ellipse geometry from a HAPY EllipsePhotometry object.

    Returns
    -------
    x0, y0, ba, pa_deg
        Center in pixels, axis ratio b/a, and PA in degrees.
    """

    import numpy as np

    if x0 is None:
        x0 = e.xcenter_best

    if y0 is None:
        y0 = e.ycenter_best

    if ba is None:
        if hasattr(e, "eps_fit") and np.isfinite(e.eps_fit):
            ba = 1.0 - e.eps_fit
        elif hasattr(e, "eps_guess") and np.isfinite(e.eps_guess):
            ba = 1.0 - e.eps_guess
        else:
            raise AttributeError("Could not find eps_fit or eps_guess to calculate b/a.")

    if pa_deg is None:
        if hasattr(e, "pa_fit") and np.isfinite(e.pa_fit):
            pa_deg = e.pa_fit
        elif hasattr(e, "pa_guess") and np.isfinite(e.pa_guess):
            pa_deg = e.pa_guess
        else:
            raise AttributeError("Could not find pa_fit or pa_guess.")

    return float(x0), float(y0), float(ba), float(pa_deg)


def get_sigma_detection_masks(
    dirname,
    sigma_H2_kpc2,
    sigma_sfr_kpc2,
    header,
    vr,
    nsigma=3,
    alma=True,
    H0=74.,
):
    """
    Create CO and Halpha detection masks on the Sigma map grid.

    Returns
    -------
    co_detected, ha_detected, detected_both
    """

    import numpy as np
    from astropy.io import fits
    from astropy import wcs
    from astropy.wcs import WCS

    vfid = dirname.split("-")[0]

    co_imwcs = WCS(header)
    kpc_per_pixel = get_kpc_per_pixel(co_imwcs, vr)
    print(f"{dirname}: kpc_per_pixel={kpc_per_pixel}")
    pixel_area_kpc2 = kpc_per_pixel**2

    D_Mpc = vr / H0
    alpha_CO = 4.3

    pixscale_deg = wcs.utils.proj_plane_pixel_scales(co_imwcs)[0]
    pixel_area_sq_deg = pixscale_deg**2

    beam_area_sq_deg = np.pi / 4 * float(header["BMAJ"]) * float(header["BMIN"])
    beam_area_pix = beam_area_sq_deg / pixel_area_sq_deg

    if alma:
        rms_map = fits.getdata(co_rms_map[vfid])

        co_rms_Msun_beam = (
            rms_map
            * 1.05e4
            * alpha_CO / 4.3
            * D_Mpc**2
        )

        co_rms_Msun_pixel = co_rms_Msun_beam / np.sqrt(beam_area_pix)
        sigma_H2_rms_kpc2 = co_rms_Msun_pixel / pixel_area_kpc2

    else:
        co_rms_Msun_beam = (
            co_rms[vfid]
            * 1.05e4
            * alpha_CO / 4.3
            * D_Mpc**2
        )

        co_rms_Msun_pixel = co_rms_Msun_beam / np.sqrt(beam_area_pix)
        sigma_H2_rms_kpc2 = co_rms_Msun_pixel / pixel_area_kpc2

    sigma_sfr_rms_kpc2 = halpha_rms[vfid] / pixel_area_kpc2

    co_detected = (
        np.isfinite(sigma_H2_kpc2)
        & (sigma_H2_kpc2 > nsigma * sigma_H2_rms_kpc2)
    )

    ha_detected = (
        np.isfinite(sigma_sfr_kpc2)
        & (sigma_sfr_kpc2 > nsigma * sigma_sfr_rms_kpc2)
    )

    detected_both = co_detected & ha_detected

    return co_detected, ha_detected, detected_both


def measure_elliptical_ks_profile_from_maps(
    sigma_H2_file,
    sigma_sfr_file,
    aperture_edges_pix,
    x0,
    y0,
    ba,
    pa_deg,
    vr,
    mask_file=None,
    dirname=None,
    nsigma=3,
    alma=True,
    require_pixel_detections=True,
    min_pixels_per_annulus=10,
):
    """
    Measure Sigma_H2 and Sigma_SFR in elliptical annuli.

    Parameters
    ----------
    sigma_H2_file : str or Path
        FITS image with Sigma_H2 in Msun/kpc^2.
    sigma_sfr_file : str or Path
        FITS image with Sigma_SFR in Msun/yr/kpc^2.
    aperture_edges_pix : array
        Outer semi-major axis values in pixels.
    x0, y0 : float
        Ellipse center in pixels.
    ba : float
        Axis ratio b/a.
    pa_deg : float
        Position angle in degrees.
    vr : float
        Cosmic recession velocity.
    mask_file : str or Path or None
        Optional mask on the same grid. 1=bad, 0=good.
    dirname : str or None
        Needed for SNR masks because it determines VFID.
    require_pixel_detections : bool
        If True, only use pixels detected at nsigma in both CO and Halpha.
        If False, use all finite positive unmasked pixels.

    Returns
    -------
    profile_df : pandas.DataFrame
        One row per annulus.
    """

    import pandas as pd
    from astropy.io import fits
    from astropy.wcs import WCS
    #from photutils.aperture import EllipticalAnnulus
    from photutils.aperture import EllipticalAnnulus, EllipticalAperture

    sigma_H2_kpc2, header = fits.getdata(sigma_H2_file, header=True)
    sigma_sfr_kpc2, sfr_header = fits.getdata(sigma_sfr_file, header=True)

    if sigma_H2_kpc2.shape != sigma_sfr_kpc2.shape:
        raise ValueError(
            f"Shape mismatch: Sigma_H2 shape={sigma_H2_kpc2.shape}, "
            f"Sigma_SFR shape={sigma_sfr_kpc2.shape}"
        )

    shape = sigma_H2_kpc2.shape

    co_imwcs = WCS(header)
    kpc_per_pixel = get_kpc_per_pixel(co_imwcs, vr)
    pixel_area_kpc2 = kpc_per_pixel**2

    if mask_file is not None:
        mask_data = fits.getdata(mask_file)
        mask_bool = mask_data > 0.5
    else:
        mask_bool = np.zeros(shape, dtype=bool)

    valid_base = (
        np.isfinite(sigma_H2_kpc2)
        & np.isfinite(sigma_sfr_kpc2)
        & (sigma_H2_kpc2 > 0)
        & (sigma_sfr_kpc2 > 0)
        & ~mask_bool
    )

    if require_pixel_detections:
        if dirname is None:
            raise ValueError("dirname must be provided when require_pixel_detections=True")

        co_detected, ha_detected, detected_both = get_sigma_detection_masks(
            dirname,
            sigma_H2_kpc2,
            sigma_sfr_kpc2,
            header,
            vr=vr,
            nsigma=nsigma,
            alma=alma,
        )

        valid_base = valid_base & detected_both

    aperture_edges_pix = np.asarray(aperture_edges_pix, dtype=float)
    aperture_edges_pix = aperture_edges_pix[np.isfinite(aperture_edges_pix)]
    aperture_edges_pix = aperture_edges_pix[aperture_edges_pix > 0]

    a_inner_edges = np.concatenate([[0.0], aperture_edges_pix[:-1]])
    a_outer_edges = aperture_edges_pix

    theta = np.deg2rad(pa_deg)

    rows = []

    for i, (a_in, a_out) in enumerate(zip(a_inner_edges, a_outer_edges)):
        if a_out <= a_in:
            continue

        b_in = ba * a_in
        b_out = ba * a_out

        if a_in <= 0:
            # central aperture
            ann = EllipticalAperture(
                positions=(x0, y0),
                a=a_out,
                b=b_out,
                theta=theta,
            )
        else:
            # true annulus
            ann = EllipticalAnnulus(
                positions=(x0, y0),
                a_in=a_in,
                a_out=a_out,
                b_in=b_in,
                b_out=b_out,
                theta=theta,
            )
        
        ann_mask_img = ann.to_mask(method="center").to_image(shape)


        if ann_mask_img is None:
            continue

        in_annulus = ann_mask_img > 0

        good = in_annulus & valid_base

        n_pix = np.sum(good)

        if n_pix < min_pixels_per_annulus:
            continue

        ##################################################
        # Pixel values
        ##################################################

        sig_H2_kpc2_pix = sigma_H2_kpc2[good]
        sig_H2_pc2_pix = sig_H2_kpc2_pix / 1e6

        sig_sfr_kpc2_pix = sigma_sfr_kpc2[good]

        log_H2_pc2_pix = np.log10(sig_H2_pc2_pix)
        log_sfr_kpc2_pix = np.log10(sig_sfr_kpc2_pix)

        log_tdep_pix = np.log10(sig_H2_kpc2_pix / sig_sfr_kpc2_pix)

        ##################################################
        # Area-weighted annular means
        ##################################################

        area_kpc2 = n_pix * pixel_area_kpc2

        H2_mass_ann = np.nansum(sig_H2_kpc2_pix * pixel_area_kpc2)
        sfr_ann = np.nansum(sig_sfr_kpc2_pix * pixel_area_kpc2)

        sigma_H2_kpc2_mean = H2_mass_ann / area_kpc2
        sigma_H2_pc2_mean = sigma_H2_kpc2_mean / 1e6

        sigma_sfr_kpc2_mean = sfr_ann / area_kpc2

        log_sigma_H2_pc2_mean = np.log10(sigma_H2_pc2_mean)
        log_sigma_sfr_kpc2_mean = np.log10(sigma_sfr_kpc2_mean)

        log_tdep_mean = np.log10(sigma_H2_kpc2_mean / sigma_sfr_kpc2_mean)

        ##################################################
        # Robust annular stats
        ##################################################

        row = {
            "annulus": i,
            "sma_in_pix": a_in,
            "sma_out_pix": a_out,
            "sma_mid_pix": 0.5 * (a_in + a_out),
            "sma_in_arcsec": a_in * np.abs(header["CDELT1"]) * 3600,
            "sma_out_arcsec": a_out * np.abs(header["CDELT1"]) * 3600,
            "sma_mid_arcsec": 0.5 * (a_in + a_out) * np.abs(header["CDELT1"]) * 3600,
            "sma_mid_kpc": 0.5 * (a_in + a_out) * kpc_per_pixel,
            "n_pix": n_pix,
            "area_kpc2": area_kpc2,

            # Area-weighted mean values
            "sigma_H2_kpc2_mean": sigma_H2_kpc2_mean,
            "sigma_H2_pc2_mean": sigma_H2_pc2_mean,
            "sigma_sfr_kpc2_mean": sigma_sfr_kpc2_mean,
            "log_sigma_H2_pc2_mean": log_sigma_H2_pc2_mean,
            "log_sigma_sfr_kpc2_mean": log_sigma_sfr_kpc2_mean,
            "log_tdep_mean": log_tdep_mean,

            # Median and 16/84 percentile ranges in log space
            "log_sigma_H2_pc2_med": np.nanmedian(log_H2_pc2_pix),
            "log_sigma_H2_pc2_p16": np.nanpercentile(log_H2_pc2_pix, 16),
            "log_sigma_H2_pc2_p84": np.nanpercentile(log_H2_pc2_pix, 84),

            "log_sigma_sfr_kpc2_med": np.nanmedian(log_sfr_kpc2_pix),
            "log_sigma_sfr_kpc2_p16": np.nanpercentile(log_sfr_kpc2_pix, 16),
            "log_sigma_sfr_kpc2_p84": np.nanpercentile(log_sfr_kpc2_pix, 84),

            "log_tdep_med": np.nanmedian(log_tdep_pix),
            "log_tdep_p16": np.nanpercentile(log_tdep_pix, 16),
            "log_tdep_p84": np.nanpercentile(log_tdep_pix, 84),
        }

        rows.append(row)

    profile_df = pd.DataFrame(rows)

    return profile_df

def get_elliptical_ks_profile_one_galaxy(
    dirname,
    vr,
    rootdir=None,
    nsigma=3,
    alma=True,
    aperture_spacing_arcsec=7.0,
    require_pixel_detections=True,
    min_pixels_per_annulus=10,
    overwrite_reproject=False,
    draw_ellipse=False,
    x0=None,
    y0=None,
    ba=None,
    pa_deg=None,
):
    """
    Measure azimuthally averaged KS quantities in R-band-defined elliptical annuli.

    Returns
    -------
    profile_df, e
        profile_df has one row per annulus.
        e is the EllipsePhotometry object used to define the annuli.
    """

    import os
    from pathlib import Path
    import numpy as np
    import matplotlib.pyplot as plt
    from astropy.io import fits
    from hapy.ellipse.photometry import EllipsePhotometry
    from photutils.aperture import EllipticalAnnulus, EllipticalAperture

    if rootdir is None:
        rootdir = os.getcwd()

    cutdir = Path(rootdir) / dirname

    r_image = cutdir / f"{dirname}-R.fits"
    r_mask = cutdir / f"{dirname}-R-mask.fits"

    sigma_H2_file = cutdir / f"{dirname}-Sigma-H2-kpc2.fits"
    sigma_sfr_file = cutdir / f"{dirname}-Sigma-sfr-kpc2.fits"

    if not sigma_H2_file.exists():
        raise FileNotFoundError(f"Missing {sigma_H2_file}")

    if not sigma_sfr_file.exists():
        raise FileNotFoundError(f"Missing {sigma_sfr_file}")

    ##################################################
    # Reproject R image and R mask to sigma/CO grid
    ##################################################

    r_co = cutdir / f"{dirname}-R-reproject-co.fits"
    r_co = reproject_to_reference(
        r_image,
        sigma_H2_file,
        r_co,
        overwrite=overwrite_reproject,
    )

    rmask_co = cutdir / f"{dirname}-R-mask-reproject-co.fits"
    rmask_co = reproject_mask_to_reference(
        r_mask,
        sigma_H2_file,
        rmask_co,
        overwrite=overwrite_reproject,
    )

    ##################################################
    # Fit R-band ellipse geometry
    ##################################################

    e = EllipsePhotometry(
        str(r_co),
        mask=str(rmask_co),
        fileid="r_ks_profile",
    )

    e.measure_sky()
    e.detect_objects()
    e.find_central_object()
    e.get_ellipse_guess()
    e.choose_best_flux_center()

    print("fit central ellipse")
    e.fit_central_ellipse()

    print("get frac masked pixels")
    e.set_guess_ellipse_mask_metrics()

    e.aperture_spacing_mode = "fixed_arcsec"
    e.aperture_spacing_arcsec = aperture_spacing_arcsec

    print("measure R-band annuli")
    e.measure_phot()
    e.calc_sb()

    if draw_ellipse:
        e.draw_guess_ellipse_mpl()
        plt.show()

    ##################################################
    # Extract geometry and measure sigma maps
    ##################################################

    x0, y0, ba, pa_deg = get_hapy_ellipse_geometry(
        e,
        x0=x0,
        y0=y0,
        ba=ba,
        pa_deg=pa_deg,
    )

    print(f"ellipse geometry: x0={x0:.1f}, y0={y0:.1f}, b/a={ba:.3f}, PA={pa_deg:.1f}")

    aperture_edges_pix = np.asarray(e.apertures_a)

    profile_df = measure_elliptical_ks_profile_from_maps(
        sigma_H2_file=sigma_H2_file,
        sigma_sfr_file=sigma_sfr_file,
        aperture_edges_pix=aperture_edges_pix,
        x0=x0,
        y0=y0,
        ba=ba,
        pa_deg=pa_deg,
        vr=vr,
        mask_file=rmask_co,
        dirname=dirname,
        nsigma=nsigma,
        alma=alma,
        require_pixel_detections=require_pixel_detections,
        min_pixels_per_annulus=min_pixels_per_annulus,
    )

    profile_df.insert(0, "dirname", dirname)
    profile_df.insert(1, "vfid", dirname.split("-")[0])

    return profile_df, e


def get_elliptical_ks_profiles_many_galaxies(
    dirnames,
    vr_dict,
    rootdir=None,
    nsigma=3,
    alma=True,
    aperture_spacing_arcsec=7.0,
    require_pixel_detections=True,
    min_pixels_per_annulus=10,
    overwrite_reproject=False,
    draw_ellipse=False,
):
    import pandas as pd

    all_profiles = []
    ellipse_objects = {}

    for dirname in dirnames:
        vfid = dirname.split("-")[0]

        if dirname in vr_dict:
            vr = vr_dict[dirname]
        elif vfid in vr_dict:
            vr = vr_dict[vfid]
        else:
            raise KeyError(f"No vr found for {dirname} or {vfid}")

        if isinstance(alma, dict):
            if dirname in alma:
                alma_flag = alma[dirname]
            elif vfid in alma:
                alma_flag = alma[vfid]
            else:
                alma_flag = False
        else:
            alma_flag = alma

        print("\n########################################")
        print(dirname)
        print("########################################")

        profile_df, e = get_elliptical_ks_profile_one_galaxy(
            dirname,
            vr=vr,
            rootdir=rootdir,
            nsigma=nsigma,
            alma=alma_flag,
            aperture_spacing_arcsec=aperture_spacing_arcsec,
            require_pixel_detections=require_pixel_detections,
            min_pixels_per_annulus=min_pixels_per_annulus,
            overwrite_reproject=overwrite_reproject,
            draw_ellipse=draw_ellipse,
        )

        all_profiles.append(profile_df)
        ellipse_objects[dirname] = e

    if len(all_profiles) == 0:
        return pd.DataFrame(), ellipse_objects

    profiles_df = pd.concat(all_profiles, ignore_index=True)

    return profiles_df, ellipse_objects
````

````python
def plot_sfe_vs_radius(
    profiles_df,
    radius_col="sma_mid_kpc",
    y_col="log_sfe_med",
    y_p16_col="log_sfe_p16",
    y_p84_col="log_sfe_p84",
    group_col="vfid",
    ax=None,
    show_fill=True,
    show_legend=True,
):
    import numpy as np
    import matplotlib.pyplot as plt

    if ax is None:
        fig, ax = plt.subplots(figsize=(6.5, 4.8))
    else:
        fig = ax.figure

    color_cycle = plt.rcParams["axes.prop_cycle"].by_key()["color"]

    for i, (gal, g) in enumerate(profiles_df.groupby(group_col)):
        g = g.sort_values(radius_col)
        color = color_cycle[i % len(color_cycle)]

        good = (
            np.isfinite(g[radius_col])
            & np.isfinite(g[y_col])
        )

        if np.sum(good) == 0:
            continue

        x = g.loc[good, radius_col]
        y = g.loc[good, y_col]

        ax.plot(
            x,
            y,
            "o-",
            color=color,
            label=str(gal),
        )

        if show_fill and y_p16_col in g.columns and y_p84_col in g.columns:
            good_fill = (
                good
                & np.isfinite(g[y_p16_col])
                & np.isfinite(g[y_p84_col])
            )

            if np.sum(good_fill) > 0:
                ax.fill_between(
                    g.loc[good_fill, radius_col],
                    g.loc[good_fill, y_p16_col],
                    g.loc[good_fill, y_p84_col],
                    color=color,
                    alpha=0.2,
                    linewidth=0,
                )

    ax.axhline(-9, color="0.6", ls="--", lw=1)
    ax.text(
        0.98,
        -9,
        r"$\tau_{\rm dep}=1\,{\rm Gyr}$",
        transform=ax.get_yaxis_transform(),
        ha="right",
        va="bottom",
        fontsize=9,
        color="0.35",
    )

    ax.set_xlabel("Semi-major axis [kpc]",fontsize=16)
    ax.set_ylabel(r"$\log_{10}\,{\rm SFE}\ [{\rm yr^{-1}}]$",fontsize=16)

    ax.grid(alpha=0.25)

    if show_legend:
        ax.legend(frameon=True, fontsize=8)

    fig.tight_layout()

    return fig, ax


def plot_annular_ks_colored_by_radius(
    profiles_df,
    ax=None,
    group_col="vfid",
    x_col="log_sigma_H2_pc2_mean",
    y_col="log_sigma_sfr_kpc2_mean",
    radius_col="sma_mid_kpc",
    cmap="viridis",
    s=60,
    show_colorbar=True,
):
    import numpy as np
    import matplotlib.pyplot as plt

    if ax is None:
        fig, ax = plt.subplots(figsize=(6.5, 5.2))
    else:
        fig = ax.figure

    markers = ["o", "s", "^", "D", "v", "P", "X", "*", "<", ">"]

    sc_last = None

    for i, (gal, g) in enumerate(profiles_df.groupby(group_col)):
        marker = markers[i % len(markers)]

        good = (
            np.isfinite(g[x_col])
            & np.isfinite(g[y_col])
            & np.isfinite(g[radius_col])
        )

        if np.sum(good) == 0:
            continue

        sc = ax.scatter(
            g.loc[good, x_col],
            g.loc[good, y_col],
            c=g.loc[good, radius_col],
            cmap=cmap,
            s=s,
            marker=marker,
            edgecolor="k",
            linewidth=0.4,
            alpha=0.9,
            label=str(gal),
            vmin=0,
            vmax=10
        )

        sc_last = sc

    # constant depletion-time lines
    xgrid = np.linspace(
        np.nanmin(profiles_df[x_col]) - 0.2,
        np.nanmax(profiles_df[x_col]) + 0.2,
        100,
    )

    for log_tdep in [8, 9, 10]:
        ygrid = xgrid + 6 - log_tdep

        ax.plot(
            xgrid,
            ygrid,
            ls="--",
            lw=1,
            color="0.6",
            alpha=0.8,
            zorder=0,
        )

        ax.text(
            xgrid[-1],
            ygrid[-1],
            rf"${10**(log_tdep-9):.1f}$ Gyr",
            fontsize=9,
            color="0.35",
            ha="right",
            va="bottom",
        )

    ax.set_xlabel(
        r"$\log_{10}\,\Sigma_{\rm H_2}\ [{\rm M_\odot\,pc^{-2}}]$", 
        fontsize=16,
    )
    ax.set_ylabel(
        r"$\log_{10}\,\Sigma_{\rm SFR}\ [{\rm M_\odot\,yr^{-1}\,kpc^{-2}}]$",
        fontsize=16
    )

    ax.grid(alpha=0.25)
    ax.legend(frameon=False, fontsize=8)

    if show_colorbar and sc_last is not None:
        cb = fig.colorbar(sc_last, ax=ax)
        cb.set_label("Semi-major axis [kpc]",size=14)

    fig.tight_layout()

    return fig, ax
````
